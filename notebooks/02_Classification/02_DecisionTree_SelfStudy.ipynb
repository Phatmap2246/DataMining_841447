```python
#Import library and create data.
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

1. Tạo dữ liệu chuột (Mice Dataset)
Đoạn code dưới đây sẽ tạo ra một file Mice_Dataset.csv với 100 con chuột, bao gồm các đặc trưng:

Weight_g: Trọng lượng (gam)

Length_cm: Chiều dài (cm)

Tail_Length_cm: Chiều dài đuôi (cm)

Fur_Color: Màu lông (Đen, Trắng, Nâu, Xám)

Diet: Chế độ ăn (Standard - Tiêu chuẩn, High-Fat - Nhiều chất béo)

Healthy: Nhãn phân loại (1 = Khỏe mạnh, 0 = Không khỏe mạnh)

```python
np.random.seed(42)
n_samples = 10

data = {
    'Weight_g': np.random.normal(loc=25, scale=5, size=n_samples), 
    'Length_cm': np.random.normal(loc=10, scale=2, size=n_samples), 
    'Tail_Length_cm': np.random.normal(loc=8, scale=1.5, size=n_samples), 
    'Fur_Color': np.random.choice(['Black', 'White', 'Brown', 'Gray'], size=n_samples), 
    'Diet': np.random.choice(['Standard', 'High-Fat'], size=n_samples) 
}

df_mice = pd.DataFrame(data)
display(df_mice.head(10))
```

```python
# phan loai suc khoe
def determine_health(row):
    is_healthy = 1
    if row['Weight_g'] < 18 or row['Weight_g'] > 32:
        is_healthy = 0
    if row['Length_cm']/row['Tail_Length_cm']>1.5:
        is_healthy = 0
    
    #make noise
    if np.random.rand()<0.1:
        is_healthy = 1 - is_healthy
    return is_healthy

df_mice['Healthy'] = df_mice.apply(determine_health, axis= 1)
df_mice
```
