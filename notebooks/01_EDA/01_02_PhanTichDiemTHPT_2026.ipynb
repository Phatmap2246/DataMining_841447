```python
thpt_math = thpt_all[['SBD','Tỉnh','Toán']]
thpt_math.head(10)
```

```python
thpt_math['Toán']=thpt_math['Toán'].fillna(-1)
thpt_math.head(10)
```

```python
math_off = thpt_math[thpt_math['Toán']==-1]
math_off.head(10)
```

```python
%matplotlib inline
dem_theo_tinh = math_off['Tỉnh'].value_counts().sort_values(ascending=False)
plt.figure(figsize=(12, 6)) # Chỉnh kích thước biểu đồ cho rộng rãi
sns.barplot(x=dem_theo_tinh.index, y=dem_theo_tinh.values, palette='viridis')

# Trang trí biểu đồ
plt.title('Số lượng thí sinh vắng mặt theo Tỉnh', fontsize=16)
plt.xlabel('Tỉnh', fontsize=12)
plt.ylabel('Số lượng thí sinh vắng', fontsize=12)
plt.xticks(rotation=45) # Xoay tên tỉnh 45 độ để không bị chồng chữ
plt.grid(axis='y', linestyle='--', alpha=0.7) # Thêm lưới ngang cho dễ nhìn

plt.tight_layout()
plt.show()
```
