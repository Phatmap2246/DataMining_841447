### 1.1.3 Tính toán các thước đo

```python
# handle with null
cusData_originall = cusData.copy()
#Fill nan with median of col
cusData['Income Level'] = cusData['Income Level'].fillna(cusData['Income Level'].median())
#Drop nan in Age col
cusData = cusData.dropna(subset=['Age'])
```

```python
#Change type of Purchase History col
cusData['Purchase History'] = pd.to_datetime(cusData['Purchase History'], format='mixed',errors='coerce')
cusData['Purchase History'].head(5)

#Create a new col 
cusData['Premium Ratio'] = cusData['Premium Amount']/cusData['Coverage Amount']
cusData.head(10)
```

# 3. Phân tích đơn biến (Univariate)

```python
#Histogram
plt.Figure(figsize=(10,5))
sns.histplot(data=cusData, x='Age', bins= 20, kde=True)
plt.title('Phân phối độ tuổi')
plt.show()
```

```python
#Boxplot (show Outler)
plt.figure(figsize=(10,5))
sns.boxplot(data=cusData,x='Income Level')
plt.title('Income Level')
plt.show()
```

```python
#Vionlin plot
plt.figure(figsize=(10,5))
sns.violinplot(data=cusData, x='Premium Amount')
plt.title('Premium Amount')
plt.show()
```
