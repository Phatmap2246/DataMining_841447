```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering
```

```python
# Tạo dữ liệu giả lập 
X,_ = make_blobs(n_samples=50, centers= 3, cluster_std=0.60, random_state=42)
plt.figure(figsize=(8, 6))

plt.scatter(X[:, 0], X[:, 1], color='blue', s=50, alpha=0.7)

plt.title("Dữ liệu ban đầu (Chưa phân cụm)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()

#Vẽ Dendrogram
plt.figure(figsize=(12,6))
plt.title("Dendrogram - Hierarchical Clustering")
# Hàm linkage tính toán khoảng cách, method='ward' là phổ biến nhất
dendrogram = sch.dendrogram(sch.linkage(X, method= 'ward'))
plt.xlabel('Data points')
plt.ylabel('Distance')
plt.axhline(y=15, color ='r', linestyle = '--', label = "Đường cắt (Threshold = 15)")
plt.legend()
plt.show()


# Áp dụng thuật toán (Cắt cây ở khoảng cách 15 sẽ ra 3 cụm)
hc = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
labels = hc.fit_predict(X)

# Show result
plt.scatter(X[:,0], X[:,1], c = labels, cmap='rainbow')
plt.title("Kết quả phân cụm (K=3)")
plt.show()
```
